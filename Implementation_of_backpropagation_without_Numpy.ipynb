```python
import numpy as np
#  Activation functions
def sigmoid(x):
    return 1 / (1 + np.exp(x))
def sigmoid_derivative(x):
    return x * (1 - x)
#  Training data
# XORlike dataset
X = np.array([[0, 0],
              [0, 1],
              [1, 0],
              [1, 1]])
y = np.array([[0],
              [1],
              [1],
              [0]])
#  Initialize weights
np.random.seed(42)
input_neurons = 2
hidden_neurons = 2
output_neurons = 1
W1 = np.random.randn(input_neurons, hidden_neurons)
b1 = np.zeros((1, hidden_neurons))
W2 = np.random.randn(hidden_neurons, output_neurons)
b2 = np.zeros((1, output_neurons))
learning_rate = 0.1
epochs = 10000
#  Training (Forward + Backpropagation)
for epoch in range(epochs):
    #  Forward propagation
    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)
    z2 = np.dot(a1, W2) + b2
    y_pred = sigmoid(z2)
    #  Loss (Mean Squared Error)
    loss = np.mean((y - y_pred) ** 2)
    #  Backpropagation
    # Output layer gradients
    d_output = (y - y_pred) * sigmoid_derivative(y_pred)
    # Hidden layer gradients
    d_hidden = np.dot(d_output, W2.T) * sigmoid_derivative(a1)
    #  Update weights and biases
    W2 += np.dot(a1.T, d_output) * learning_rate
    b2 += np.sum(d_output, axis=0, keepdims=True) * learning_rate
    W1 += np.dot(X.T, d_hidden) * learning_rate
    b1 += np.sum(d_hidden, axis=0, keepdims=True) * learning_rate
    #  Print loss occasionally
    if epoch % 2000 == 0:
        print(f"Epoch {epoch}, Loss: {loss:.4f}")
#  Testing
print("\nFinal Predictions:")
print(y_pred.round(3))
```

```python
import math
import random
# Sigmoid function
def sigmoid(x):
    return 1 / (1 + math.exp(-x))

# Derivative of sigmoid
def sigmoid_derivative(y):
    return y * (1 - y)

# Training data (XOR)
X = [[0,0], [0,1], [1,0], [1,1]]
Y = [0, 1, 1, 0]

# Initialize weights and biases randomly
w1 = [[random.random(), random.random()],
      [random.random(), random.random()]]
b1 = [0, 0]

w2 = [random.random(), random.random()]
b2 = random.random()

learning_rate = 0.5

# Training
for epoch in range(10000):
    total_error = 0

    for x, y in zip(X, Y):

        # -------- Forward Pass --------
        h = []
        for i in range(2):
            net = x[0]*w1[0][i] + x[1]*w1[1][i] + b1[i]
            h.append(sigmoid(net))

        out_net = h[0]*w2[0] + h[1]*w2[1] + b2
        output = sigmoid(out_net)

        # Error
        error = y - output
        total_error += error**2

        # -------- Backpropagation --------
        d_out = error * sigmoid_derivative(output)

        d_h = []
        for i in range(2):
            d_h.append(d_out * w2[i] * sigmoid_derivative(h[i]))

        # Update weights (hidden → output)
        for i in range(2):
            w2[i] += learning_rate * d_out * h[i]
        b2 += learning_rate * d_out

        # Update weights (input → hidden)
        for i in range(2):
            w1[0][i] += learning_rate * d_h[i] * x[0]
            w1[1][i] += learning_rate * d_h[i] * x[1]
            b1[i] += learning_rate * d_h[i]

    if epoch % 1000 == 0:
        print("Epoch:", epoch, "Error:", total_error)

# -------- Final Output --------
print("\nFinal Predictions:")
for x in X:
    h = []
    for i in range(2):
        net = x[0]*w1[0][i] + x[1]*w1[1][i] + b1[i]
        h.append(sigmoid(net))

    out_net = h[0]*w2[0] + h[1]*w2[1] + b2
    print(x, "->", round(sigmoid(out_net), 3))











```
